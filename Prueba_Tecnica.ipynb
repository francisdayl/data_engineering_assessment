```python
# Ejercicio # 1
conn = sqlite3.connect('test.db')


####PROGRAME SU RESPUESTA DENTRO DEL CONN.EXECUTE ENTRE LAS COMILLAS

cursor = conn.execute('''
WITH partidos_cleaned AS (
  SELECT REPLACE(
           REPLACE(
             REPLACE(
               REPLACE(
                 REPLACE(
                   REPLACE(TRIM(UPPER(partido),'.'),
                   'á', 'A'),
                 'é', 'E'),
               'í', 'I'),
             'ó', 'O'),
           'ú', 'U'),
          'ñ', 'N') AS partido_cleaned, 
          grupo, 
          resultado
  FROM partidos
),
partidos_expandidos AS (
  SELECT
    TRIM(SUBSTR(partido_cleaned, 1, INSTR(partido_cleaned, 'VS.') - 1)) AS equipo, grupo,
    TRIM(SUBSTR(partido_cleaned, INSTR(partido_cleaned, 'VS.') + 3)) AS oponente,
    CAST(SUBSTR(resultado, 1, INSTR(resultado, ',') - 1) AS INTEGER) AS goles_favor,
    CAST(SUBSTR(resultado, INSTR(resultado, ',') + 1) AS INTEGER) AS goles_contra
  FROM partidos_cleaned

  UNION ALL

  SELECT
    TRIM(SUBSTR(partido_cleaned, INSTR(partido_cleaned, 'VS.') + 3)) AS equipo,grupo,
    TRIM(SUBSTR(partido_cleaned, 1, INSTR(partido_cleaned, 'VS.') - 1)) AS oponente,
    CAST(SUBSTR(resultado, INSTR(resultado, ',') + 1) AS INTEGER) AS goles_favor,
    CAST(SUBSTR(resultado, 1, INSTR(resultado, ',') - 1) AS INTEGER) AS goles_contra
  FROM partidos_cleaned
),
partidos_con_puntos AS (
  SELECT
    equipo,
    grupo,
    goles_favor,
    goles_contra,
    CASE
      WHEN goles_favor > goles_contra THEN 3
      WHEN goles_favor = goles_contra THEN 1
      ELSE 0
    END AS puntos,
    goles_favor - goles_contra AS gol_diferencia
  FROM partidos_expandidos
)


SELECT
  equipo,
  grupo,
  SUM(puntos) AS puntos,
  SUM(gol_diferencia) AS gol_diferencia
FROM partidos_con_puntos
GROUP BY equipo
ORDER BY grupo, equipo;

                      ''')

for row in cursor:
  print(row)
conn.close()
```

```python
# Ejercicio # 2
conn = sqlite3.connect('test.db')

####PROGRAME SU RESPUESTA DENTRO DEL CONN.EXECUTE ENTRE LAS COMILLAS

cursor = conn.execute('''
WITH ranked_teams AS (
  SELECT 
    equipo,
    grupo,
    ROW_NUMBER() OVER (
      PARTITION BY grupo 
      ORDER BY puntos DESC, gol_diferencia DESC
    ) AS ranking_grupal
  FROM Resultados_Qatar rq
  
),
equipos_clasificados AS (
	SELECT 
	  ranking_grupal,
	  equipo,
	  grupo
	FROM ranked_teams
	WHERE ranking_grupal <= 2
	ORDER BY grupo, ranking_grupal
),
primeros_clasificados AS (
	SELECT ROW_NUMBER() OVER () AS idx, * 
	FROM equipos_clasificados
	WHERE ranking_grupal = 1
),
segundos_clasificados AS (
	SELECT ROW_NUMBER() OVER () AS idx, * 
	FROM equipos_clasificados
	WHERE ranking_grupal = 2
),
encuentros AS (
	SELECT 
		pc.ranking_grupal as primeros_ranking,
		pc.equipo as primeros_equipo,
		pc.grupo as primeros_grupo,
		sc.ranking_grupal as segundos_ranking,
		sc.equipo as segundos_equipo,
		sc.grupo as segundos_grupo
	FROM primeros_clasificados pc
	INNER JOIN segundos_clasificados sc ON (sc.idx = CASE 
		WHEN pc.idx % 2 = 1 THEN pc.idx + 1 
	    ELSE pc.idx - 1 
	END
	)
)
SELECT
	e.primeros_equipo || ' vs. ' || e.segundos_equipo as encuentro,
	c.fecha,
	c.hora ,
	c.sede
FROM
	encuentros e
INNER JOIN Clasificados c ON
	UPPER(c.partido) LIKE ('%1º DEL ' || e.primeros_grupo ||
	' vs. 2º del ' || e.segundos_grupo
	|| '%');



''')

for row in cursor:
  print(row)
conn.close()
```

```python
# Respuesta 5.2
df = spark.read.option("multiLine", True).json("CENTROS_EDUCATIVOS_MADRID.json")

df = df.withColumn("direccion_coor_x", col("direccion_coor_x").cast("integer"))
df = df.withColumn("direccion_coor_y", col("direccion_coor_y").cast("integer"))

df.show(truncate=False)
```

```python
# Respuesta 5.2.1

df_avgs = df.groupBy("centro_tipo_descripcion").agg(
   round( avg("direccion_coor_x"),4).alias("avg_coor_x"),
    round(avg("direccion_coor_y"),4).alias("avg_coor_y")
)

df_avgs.show(truncate=False)
```

```python
# Respuesta 5.2.2
import math
def euclidean_distance(x, y, x0, y0):
    if x is None or y is None:
        return None
    return math.sqrt((x - x0)**2 + (y - y0)**2)

euclidean_udf = udf(euclidean_distance, DoubleType())
df_with_avgs = df.join(
    df_avgs,
    on="centro_tipo_descripcion",
    how="inner"
)

df_with_euclidian_values = df_with_avgs.withColumn("distancia_origen", 
                                         euclidean_udf(
                                             col("direccion_coor_x"), 
                                             col("direccion_coor_y"), 
                                             col("avg_coor_x"), 
                                             col("avg_coor_y"))
                                        )
df_with_euclidian_values.show(truncate=False)
```

```python
# Respuesta 5.2.3

window_dist = Window.partitionBy("centro_tipo_descripcion").orderBy(col("distancia_origen").asc())

df_ranked_distance= df_with_euclidian_values.withColumn(
    "ranking_distancia",
   rank().over(window_dist)
)
df_distance_result = df_ranked_distance.withColumn(
    "mas_cercano", 
    when(col("ranking_distancia") == 1, lit(True)).otherwise(lit(False))
)
df_distance_result.where(col("mas_cercano")==True).show(truncate=False)
```
